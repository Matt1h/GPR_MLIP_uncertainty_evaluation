```python
from os.path import join
import torch
import matplotlib.pyplot as plt

from GPR_MLIP.evaluation import extended_reliablility_diagram, plot_legend

RESULT_ROOT_PATH = join('..', 'results', 'uncertainty_error_calculation')
```

```python
def plot_correlations_ensemble_vs_single(
        axes,
        uncertainties_,
        errors_,
        max_values_theory,
        **kwargs
):
    # Two sets
    extended_reliablility_diagram(
        axes[0],
        uncertainties_['Two sets']*1000,
        errors_['GPR model']*1000,
        max_value_theory=max_values_theory[0],
        **kwargs,
    )
    axes[0].set_title('Two sets single model', fontsize=16)

    extended_reliablility_diagram(
        axes[1],
        uncertainties_['Two sets']*1000,
        errors_['Two sets']*1000,
        max_value_theory=max_values_theory[0],
        **kwargs,
    )
    axes[1].set_title('Two sets ensemble', fontsize=16)

    # Bootstrap
    extended_reliablility_diagram(
        axes[2],
        uncertainties_['Bootstrap aggregation']*1000,
        errors_['GPR model']*1000,
        max_value_theory=max_values_theory[1],
        **kwargs,
    )
    axes[2].set_title('Bootstrap aggregation single model', fontsize=16)

    extended_reliablility_diagram(
        axes[3],
        uncertainties_['Bootstrap aggregation']*1000,
        errors_['Bootstrap aggregation']*1000,
        max_value_theory=max_values_theory[1],
        **kwargs,
    )
    axes[3].set_title('Bootstrap aggregation ensemble', fontsize=16)
```

## rMD17 Benzene

### Coulomb

```python
uncertainties = torch.load(
    join(RESULT_ROOT_PATH, 'rmd17', 'benzene', 'GPR_coulomb_rbf', 'uncertainties'),
    weights_only=True
)
errors = torch.load(
    join(RESULT_ROOT_PATH, 'rmd17', 'benzene', 'GPR_coulomb_rbf', 'errors'),
    weights_only=True
)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))

plot_correlations_ensemble_vs_single(
    axes, 
    uncertainties, 
    errors,
    max_values_theory=[25, 15],
    max_value=30,
    ticks=[0, 15, 30],
    bin_size=1,
)

plot_legend(axes[0], fontsize=14, loc='lower right')
plt.tight_layout()
```

```python
uncertainties = torch.load(
    join(RESULT_ROOT_PATH, 'rmd17', 'benzene', 'GPR_soap_atomistic_atomistic_sum1', 'uncertainties'),
    weights_only=True
)
errors = torch.load(
    join(RESULT_ROOT_PATH, 'rmd17', 'benzene', 'GPR_soap_atomistic_atomistic_sum1', 'errors'),
    weights_only=True
)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))

plot_correlations_ensemble_vs_single(
    axes, 
    uncertainties, 
    errors,
    max_values_theory=[25, 15],
    max_value=30,
    ticks=[0, 15, 30],
    bin_size=1,
)

plot_legend(axes[0], fontsize=14, loc='lower right')
plt.tight_layout()
```
